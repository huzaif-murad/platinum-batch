#### Radix Sort

```python
my_list=[123,321,12,43,456,122,455,65]
largest=max(my_list)
place=1

while(place<=largest):
    bucket=[[] for _ in range(10)]

    for num in my_list:
        key=(num//place)%10
        bucket[key].append(num)
    my_list=[]
    my_list.extend(bucket)
    print(my_list)
    place*=10
     
```
